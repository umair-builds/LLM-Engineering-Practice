```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
import gradio as gr # oh yeah!

# Load environment variables from .env file
load_dotenv()
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama" # We don't actually need a key for local, but the library asks for one
)
```

```python
message = "I'm a student"
messages = [{"role": "user", "content": message}]
messages
```

```python
response = client.chat.completions.create( model="llama3.2:1b", messages=messages,)

response.choices[0].message.content
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')
groq_url = "https://api.groq.com/openai/v1"
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
```

```python
if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")
```

```python
def Groq(messages): 
    response = groq.chat.completions.create( model="openai/gpt-oss-120b", messages=messages, max_tokens=100)
    display(Markdown(response.choices[0].message.content))
```

```python
gr.Interface(fn=groq, inputs="textbox", outputs="textbox", flagging_mode="never").launch()
```
