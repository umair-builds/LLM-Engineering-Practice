```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
import gradio as gr # oh yeah!

# Load environment variables from .env file
load_dotenv()
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama" # We don't actually need a key for local, but the library asks for one
)
```

```python
message = "I'm a student"
messages = [{"role": "user", "content": message}]
messages
```

```python
response = client.chat.completions.create( model="llama3.2:1b", messages=messages,)

response.choices[0].message.content
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')
groq_url = "https://api.groq.com/openai/v1"
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
```

```python
if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")
```

```python
system_message = "You are a helpful assistant"
def Groq(prompt): 
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = groq.chat.completions.create( model="openai/gpt-oss-120b", messages=messages, max_tokens=100)
    return response.choices[0].message.content
```

```python
def shout(text):
    print(f"Shout has been called with input {text}")
    return text.upper()
```

```python
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch()
```

```python
# Adding share=True means that it can be accessed publically
# A more permanent hosting is available using a platform called Spaces from HuggingFace, which we will touch on next week
# NOTE: Some Anti-virus software and Corporate Firewalls might not like you using share=True. 
# If you're at work on on a work network, I suggest skip this test.

gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch(share=True)
```

```python
# Adding inbrowser=True opens up a new browser window automatically

gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch(inbrowser=True)
```

```python
# Adding a little more:

from tkinter import TRUE


message_input = gr.Textbox(label="Your message:", info="Enter a message to be shouted", lines=7)
message_output = gr.Textbox(label="Response:", lines=8)

view = gr.Interface(
    fn=Groq,
    title="Groq", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=["Hi, I'm Umair", "Looking for a Urgent help"], 
    flagging_mode="never"
    )
view.launch(inbrowser=TRUE)
```
