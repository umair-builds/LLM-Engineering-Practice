## Take a look to Groq

```python
# imports

import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

# Load environment variables from .env file
load_dotenv()
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')
groq_url = "https://api.groq.com/openai/v1"
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
```

```python
if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")
```

```python
tell_a_joke = [
    {"role": "user", "content": "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"},
]
```

```python
response = groq.chat.completions.create(model="openai/gpt-oss-120b", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```

Other way to use groq

```python
from groq import Groq

client = Groq(api_key=groq_api_key)
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=tell_a_joke,
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

```python
easy_puzzle = [
    {"role": "user", "content": 
        "You toss 2 coins. One of them is heads. What's the probability the other is tails? Answer with the probability only."},
]
```

```python
response = groq.chat.completions.create(model="openai/gpt-oss-120b", messages=easy_puzzle)
display(Markdown(response.choices[0].message.content))
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama" # We don't actually need a key for local, but the library asks for one
)

response = client.chat.completions.create(
    model="llama3.2:1b",
    messages=easy_puzzle,
)

display(Markdown(response.choices[0].message.content))
```
